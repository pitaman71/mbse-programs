# Suite TRN — Translating TypeScript to Python

`Transpilers.TypeScriptToPython` translates a TypeScript 5 program the store holds into the transaction that writes a
Python 3 module into the same store, which never passes through text: `transpile(store, path, target)` reads the
program at `path` (`TypeScript5`'s tree, values read record by record, the statements within a function's term among
them) and gives the transaction's callback, which writes the module at `target` as `Python3.Parse` writes one, its
tree one nested value built through the builders, its expressions `Python3` terms built by the dialect's own writers.
`Transpilers.store()` holds both languages' schemas. Each case parses its program into one store
(`TypeScript5.Parse.parse`), transpiles it, and checks the text the module prints to (`Python3.Print`). The programs of
`conformance/transpilers/typescript-to-python` also run, and print what `<program>.out` holds: here their translation
runs in CPython. The cases port 0.3's suite TRN (`14_TypeScriptToPython.ipynb`), its expected texts kept; what does
not translate fails at its syntax node's anchor (`TypeScript5.Definitions.Anchor`), where 0.3 gave its path.

```python
import contextlib
import io
import itertools
from pathlib import Path

from mbse.Expressions.Framework import Objects, Terms
from mbse.Programs import Python3, Transpilers, TypeScript5
from mbse.Programs.Conformance.write import transpiled
from mbse.Programs.Framework.Errors import TranspileError
from mbse.Programs.Framework.Model import WORKSPACE
from mbse.Programs.Transpilers.TypeScriptToPython import transpile, transpile_file
from mbse.Schemas.Framework import Plain
from support import raises

PROGRAMS = Path("../../conformance/transpilers/typescript-to-python")
STORE = Transpilers.store()
PATHS = (f"case{n}" for n in itertools.count())


def parsed(text):
    """The path of the TypeScript program `text` parses to, written in the store at a path of its own."""
    path = next(PATHS) + ".ts"
    STORE.transaction(TypeScript5.Parse.parse(text, path))
    return path


def translate(text):
    """The text of the Python module the program `text` translates to, written beside it."""
    path = parsed(text)
    target = path[:-3] + ".py"
    STORE.transaction(transpile(STORE, path, target))
    return Python3.Print.print_module(STORE, target).rstrip("\n")


def translates(ts, py):
    assert translate(ts) == py, translate(ts)


def fails(ts, message):
    path = parsed(ts)
    with raises(TranspileError, match=message):
        transpile(STORE, path, path[:-3] + ".py")
```

<a id="TRN-01"></a>
## TRN-01 · Programs keep their meaning: each translation is current, and prints what the TypeScript program prints

`Conformance.write.transpiled` translates each program as the conformance writer does, in a store of its own, and its
text is the `.py` beside it; run, it prints what the `.out` beside it holds.

```python
files = transpiled(PROGRAMS)
for program in sorted(PROGRAMS.glob("*.ts")):
    python = (PROGRAMS / f"{program.stem}.py").read_text(encoding="utf-8")
    assert files[f"{program.stem}.py"] == python, f"{program.stem}.py is stale"
    out = io.StringIO()
    with contextlib.redirect_stdout(out):
        exec(compile(python, program.stem, "exec"), {"__name__": "__main__"})
    assert out.getvalue() == (PROGRAMS / f"{program.stem}.out").read_text(encoding="utf-8"), out.getvalue()
print(len(files), "programs")
```

## TRN-08 · The transaction: read in one, written in another, one upsert of the module's tree as one value

`transpile` reads the program within a transaction of its own, LookupError where the workspace holds no file at its
path; `transpile_file` reads a visitor within the caller's. What either gives is the callback of the transaction that
writes the module: plain specs, holding no visitor, which outlive the transaction that read the program, executed by
`store.transaction` or built by `store.Mutations.OfTransaction`, one upsert, `uid(target)`, as `Python3.Parse` gives one.
The module is a `Programs.Python3.Module` the workspace owns at the target's path, its statements values with no
line, as built rather than parsed, and valid (`Python3.Check.validate`), each program's among them.
`Transpilers.store()` runs both languages' installs, each once.

```python
assert len(set(Transpilers.INSTALLS)) == len(Transpilers.INSTALLS)
assert set(TypeScript5.INSTALLS) | set(Python3.INSTALLS) == set(Transpilers.INSTALLS)
with raises(LookupError):
    transpile(STORE, "absent.ts", "absent.py")

path = parsed("function f(n: number) { return n + 1; }\nconsole.log(f(1));\n")
with STORE.transaction():
    callback = transpile_file(STORE, STORE.singleton(WORKSPACE).files({"path": path}), "f.py")
built = Plain.ToPlain(STORE).Transaction(STORE.Mutations.OfTransaction(callback))
assert built["uid"] == "f.py" and [list(m) for m in built["mutations"]] == [["upsert"]]
STORE.transaction(callback)
with STORE.transaction():
    module = STORE.singleton(WORKSPACE).files({"path": "f.py"})
    assert module.as_schema().name == "Programs.Python3.Module"
    statements = [Terms.kind_of(s) for s in Objects.properties(module)["body"].values]
    assert [tag for tag, _ in statements] == ["functiondef", "expr"]
    assert all("line" not in Objects.properties(record) for _, record in statements)
assert Python3.Print.print_module(STORE, "f.py") == "def f(n: float):\n    return n + 1\n\n\nprint(f(1))\n"
assert Python3.Check.validate(STORE, "f.py") == []

for program in sorted(PROGRAMS.glob("*.ts")):
    path = parsed(program.read_text(encoding="utf-8"))
    STORE.transaction(transpile(STORE, path, f"{program.stem}.py"))
    assert Python3.Check.validate(STORE, f"{program.stem}.py") == [], program.stem
print("ok")
```
